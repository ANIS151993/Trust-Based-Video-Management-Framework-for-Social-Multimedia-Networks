# Trust-Based Video Management Framework — Validation Lab

Runnable companion notebook for the paper *"Trust-Based Video Management Framework for Social Multimedia Networks"* by Md Anisur Rahman Chowdhury & Samuel Tweneboah-Koduah.

This notebook:
1. Generates a synthetic 5-class silhouette-proxy dataset (no real weapon imagery).
2. Trains VGG16 and ResNet50 via transfer learning, exactly as described in Section III-C of the paper.
3. Runs the dynamic trust-scoring simulation (Section III-B).
4. Produces all figures used in the paper and on the project website.

**Runtime:** a few minutes on CPU with the default (quick) settings.
